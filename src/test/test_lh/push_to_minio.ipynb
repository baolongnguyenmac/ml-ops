# Push data to MinIO

- Prerequire: Obtain 3 `parquet` files in `../data` from `./generate_data.ipynb`

```python
from minio import Minio
import os
from pathlib import Path
```

```python
client = Minio(
    endpoint='localhost:9099',
    access_key='minio_access_key',
    secret_key='minio_secret_key',
    secure=False,
)
```

```python
bucket_name = 'store-lh'
found = client.bucket_exists(bucket_name=bucket_name)
if not found:
    client.make_bucket(bucket_name=bucket_name)
else:
    print(f'Bucket {bucket_name} already exists, skip creating!')
```

```python
bla = '../../../data/test/lh'
len(bla)
```

```python
data_path = Path('../../../data/test/lh')

for fp in list(data_path.rglob('*.parquet')) + list(data_path.rglob('*.json')):
    print(f"Uploading {fp}")
    client.fput_object(
        bucket_name=bucket_name,
        object_name=fp._str[len('../../../data/test/lh/'):], # remove the ../data/store/ at the beginning of the path
        file_path=fp
    )
```
