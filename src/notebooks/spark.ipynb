- Thử tạo 1 `tmp_df` có các cột chứa các kiểu dữ liệu nêu trên

````python
from pyspark.sql import DataFrame
from pyspark.sql.types import StructType, StructField, StringType, FloatType, DoubleType, TimestampType, IntegerType
from datetime import datetime, timedelta
import random

# schema = StructType([
#     StructField("str_col", StringType(), True),
#     StructField("float_col", FloatType(), True),
#     StructField("double_col", DoubleType(), True),
#     StructField("ts_col", TimestampType(), True),
#     StructField("int_col", IntegerType(), True),
# ])

schema = ["str_col", "float_col", "double_col", "ts_col", "int_col"]

base_time = datetime(2024, 1, 1)
data = [
    (
        f"name_{i}",
        round(random.uniform(0, 100), 2),
        random.uniform(0, 1000),
        base_time + timedelta(hours=i),
        random.randint(0, 1000),
    )
    for i in range(100)
]

# hàm này nhận vào list các rows (value only), cần có schema
# hoặc này nhận vào list các rows (dict[col_name, value]), không cần schema
tmp_df: DataFrame = spark.createDataFrame(data, schema=schema)
tmp_df.show(5)
tmp_df.printSchema()
````

### Handle missing data

````python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('Handle missing data').getOrCreate()
spark
````

````python
df_spark = spark.read.csv('../../data/test/csv/04.customer.csv', header=True, inferSchema=True)
df_spark.show(2)
````

````python
df_spark.describe().show()
````

````python
df_spark.printSchema()
````

- Khi `.describe()` và `.printSchema()`
    - `describe` sẽ chạy thống kê cho một cột theo kiểu dữ liệu phổ biến nhất. Nghĩa là nếu trong 1 cột toàn float có chứa 1 giá trị string thì phần thống kê sẽ bỏ giá trị đó ra (cho các tác vụ liên quan đến tính toán)
    - `printSchema` sẽ in ra kiểu dữ liệu theo common logic, nghĩa là chọn kiểu dữ liệu tổng quát. Ví dụ 1 cột toàn float có chứa 1 giá trị string thì kiểu dữ liệu của cột sẽ là string

- Detect `NULL` value. Trong `pyspark` thì `NULL` và `NaN` là khác nhau. `NaN` là 1 giá trị được thiết kế cho kiểu số, `NULL` là trạng thái chung cho cả kiểu số và non-số
- Khi kiểu số chứa giá trị `NaN`, nó không `NULL`

````python
[{col: df_spark.filter(df_spark[f'{col}'].isNull()).count()} for col in df_spark.columns]
````

````python
df_spark.count()
````

- Xoá dòng có dữ liệu trống trong df

````python
df_spark.dropna(how='any').count()

# xoá kiểu này thì bay hết dữ liệu --> chỉ xoá những dòng có những cột thiếu nhất định (bằng cách lọc các dòng không bị xoá ra 1 df khác)
````

````python
df_spark.filter(~df_spark['salesRepEmployeeNumber'].isNull() | ~df_spark['state'].isNull()).count()
````

````python
# xoá những dòng có ít hơn `thresh` giá trị real
# hay có nhiều hơn len(df_spark.columns)-thresh giá trị null
# trong ví dụ này: Xoá các dòng có nhiều hơn 13-(13-2) = 2 giá trị null
df_spark.dropna(how='any', thresh=len(df_spark.columns)-2).count()
````

- Có thể xoá trên 1 subset các cột

````python
# 122 - 7 = 115
df_spark.dropna(how='any', subset=['postalCode']).count()
````

- Fill nan

````python
from pyspark.sql import functions as F

# collect() returns a list of rows
mean_value = df_spark.select(F.mean('customerNumber')).collect()[0][0]
df_spark.fillna(mean_value, subset=['customerNumber']).show()
````

````python
# fill nan with median
median_value = df_spark.select(F.median('customerNumber')).collect()[0][0]
df_spark.fillna(median_value, subset=['customerNumber']).show()
````

### Apply a function to a column

````python
from pyspark.sql.types import StringType

def create_fullname(first_name:str, last_name:str):
    return first_name + ' ' + last_name

create_fullname_udf = F.udf(create_fullname, StringType())

df_spark.withColumn('Full name', create_fullname_udf(df_spark['contactLastName'], df_spark['contactFirstName'])).show()
````

### Group by & Aggregate

````python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder.appName('agg').getOrCreate()
spark
````

````python
df_spark = spark.read.csv('../../data/test/csv/04.customer.csv', header=True, inferSchema=True)
df_spark.show(5)
````

````python
# count the number of order in every country/state/city
df_orderByCountry = df_spark.groupBy('country').count()
df_orderByCountry.select(F.sum(df_orderByCountry['count'])).show()
````

````python
df_spark.groupBy('country').agg(F.max(df_spark['creditLimit'])).show()
````

````python
%%timeit
df_spark.agg({'customerNumber': 'sum'}).show()
````

````python
%%timeit
df_spark.select(F.sum('customerNumber')).show()
````

- Có vẻ việc dùng hàm sẽ ổn định hơn và nhanh hơn 1 chút so với việc truyền vào key, value cho `agg()` tự parse

````python
# người có tín dụng cao nhất tại mỗi thành phố của Mỹ
maxCredByCity_df = df_spark.filter(df_spark.country=='USA').groupBy('city').agg(F.max('creditLimit')).alias('max_df')
maxCredByCity_df.join(
    df_spark.alias('orig_df'),
    (F.col('max_df.city') == F.col('orig_df.city')) & (maxCredByCity_df['max(creditLimit)'] == df_spark['creditLimit']),
    'inner'
)['max_df.city', 'customerName', 'max(creditLimit)'].show()
````
